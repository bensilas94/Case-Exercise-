# Physics-Driven Parameter Mapping & ML Feature Engineering for Subsea XT FMECA

To transform our static FMECA sheet into a predictive health monitoring or Prognostics and Health Management (PHM) system, we must map the qualitative failure modes to their underlying physical parameters and construct relevant machine learning features.

---

### 1. Production Master Valve (PMV) — Fail to close on demand
*   **Driving Physics Parameters:**
    *   *Fluid Dynamics & Tribology:* Friction coefficient between gate and seats, hydraulic actuation pressure ($P_{hyd}$), spring mechanical return force ($F_s = k \cdot x$), and fluid shear stress.
    *   *Thermodynamics:* High-temperature scale precipitation kinetics.
*   **ML Feature Engineering Strategies:**
    *   **Hydraulic Signature Profile:** Integral of hydraulic pressure curve over actuation time ($∫ P_{hyd} \, dt$) to quantify actuation energy change.
    *   **Actuation Time Deviation:** Time difference between control signal emission and limit switch transition ($\Delta t_{act}$).
    *   **Scale Accumulation Indicator:** Cumulative flow volume multiplied by production fluid temperature ($V_{cum} \times T_{avg}$).

### 2. Production Wing Valve (PWV) — External hydraulic fluid leakage
*   **Driving Physics Parameters:**
    *   *Fluid Mechanics:* Leakage rate through seals ($Q = \frac{\pi d c^3 \Delta P}{12 \mu L}$ where $c$ is clearance, $\mu$ is viscosity).
    *   *Materials Science:* Viscoelastic degradation of elastomer seals under high sea-bottom pressure.
*   **ML Feature Engineering Strategies:**
    *   **Pressure Decay Rate:** Rate of pressure drop in the hydraulic line during lock-in states ($\frac{dP_{hyd}}{dt}$).
    *   **Fluid Consumption Anomalies:** Cumulative supply pump run-time versus expected theoretical stroke volume.
    *   **Viscosity Correction Index:** Normalization of hydraulic pressure response based on sea-ambient temperature.

### 3. Subsea Control Module (SCM) — Complete loss of communication
*   **Driving Physics Parameters:**
    *   *Electromagnetics & Electrochemistry:* Dielectric breakdown of seawater inside connectors, galvanic corrosion, cable insulation resistance degradation ($R_{ins}$).
*   **ML Feature Engineering Strategies:**
    *   **Time-Domain Reflectometry (TDR) Features:** Wave reflection amplitude anomalies to pinpoint cable degradation locations.
    *   **Signal-to-Noise Ratio (SNR) Trends:** Rolling average of communication channel SNR and packet drop rate.
    *   **Leakage Current Metrics:** Ground isolation resistance ($M\Omega$) tracking over time.

### 4. Wellhead Connector — Gasket sealing failure
*   **Driving Physics Parameters:**
    *   *Structural Mechanics:* Contact stress profile on the VX/VT gasket, tension/bending moments on the wellhead housing due to thermal expansion and environmental loads.
*   **ML Feature Engineering Strategies:**
    *   **Dynamic Load Indices:** Frequency domain features (FFT amplitude peaks) of structural vibration from acoustic or strain sensors during cyclic wave/current loading.
    *   **Thermal Expansion Delta:** $\Delta T$ between the internal process fluid and external seawater, indicating thermal stress on the metal-to-metal seal.

### 5. Choke Valve — Severe internal erosion
*   **Driving Physics Parameters:**
    *   *Multiphase Fluid Dynamics:* Sand particle impact kinetics, liquid/gas velocity profiles, erosion rate ($E \propto v^n$ where $v$ is fluid velocity and $n \approx 2-3$).
*   **ML Feature Engineering Strategies:**
    *   **Acoustic Sand Energy Index:** Integral of high-frequency acoustic sensor signals dedicated to sand detection.
    *   **Flow Coefficient (Cv) Deviation:** Discrepancy between theoretical $C_v$ (based on choke opening % and $\Delta P$) and actual measured flow rate.
    *   **Kinetic Energy Proxy:** Product of fluid density, velocity squared, and calculated sand concentration ($KE_{proxy} = \rho \cdot v^2 \cdot C_{sand}$).

### 6. Pressure & Temperature Sensor (PT) — Signal drift
*   **Driving Physics Parameters:**
    *   *Solid-State Physics:* Piezoresistive sensor element aging, thermal hysteresis, signal conversion circuit component drift.
*   **ML Feature Engineering Strategies:**
    *   **Spatial Consistency Residuals:** Cross-sensor difference metrics comparing adjacent sensors (e.g., $|P_{XT\_bore} - P_{manifold}|$).
    *   **Transient Response Alignment:** Phase lag/correlation parameters between temperature and pressure during sudden well shut-ins.

### 7. Chemical Injection Valve (CIV) — Internal blockage / plugged nozzle
*   **Driving Physics Parameters:**
    *   *Thermodynamics & Chemistry:* Phase envelope boundaries for hydrates, mineral solubility limits as pressure drops across the nozzle.
*   **ML Feature Engineering Strategies:**
    *   **Injection Line Flow Resistance:** Calculated hydraulic resistance $R = \frac{\Delta P_{injection}}{Q_{chemical}^2}$.
    *   **Sub-cooling Margin:** Difference between ambient subsea temperature and hydrate formation temperature threshold at current operating pressure ($T_{margin} = T_{ambient} - T_{hydrate}(P)$).

### 8. SCSSV Control Line — Hydraulic line blockage
*   **Driving Physics Parameters:**
    *   *Fluid Mechanics:* Non-Newtonian behavior of contaminated hydraulic fluids, particulate settlement mechanics (Stokes' Law).
*   **ML Feature Engineering Strategies:**
    *   **Transient Pressurization Slope:** Initial derivative of pressure build-up when pump starts ($│\frac{dP}{dt}\u2502_{t=0}$).
    *   **Viscous Dampening Coefficient:** Curve-fitting parameters of the hydraulic system's step-response decay curve.

### 9. Tree Piping / Cross-Over Valve (XOV) — Internal passing
*   **Driving Physics Parameters:**
    *   *Fluid Dynamics:* Orifice flow under micro-clearances, acoustic emission generated by turbulent gas/liquid leak paths.
*   **ML Feature Engineering Strategies:**
    *   **A-B Annulus Pressure Build-up Slope:** Rate of pressure rise in the closed annulus during normal production.
    *   **Acoustic Emission (AE) RMS:** Root Mean Square of high-frequency structural acoustic signals around the valve body.

### 10. Tree Cap — Corrosion protection failure / anode depletion
*   **Driving Physics Parameters:**
    *   *Electrochemistry:* Galvanic cell current density between steel and sacrificial zinc/aluminum anodes, oxygen concentration gradients.
*   **ML Feature Engineering Strategies:**
    *   **Cathodic Protection (CP) Potential:** ROV-measured electrical potential field strength normalized by salinity.
    *   **Cumulative Environmental Exposure:** Integrated time of structural exposure to corrosive environment, weighted by local seawater flow velocity.

Question 3.Implement at least two of the physics models in Python and validate against hand calculations.


```python
import numpy as np

print("=== PHYSICS MODEL 1: CHOKE VALVE INTERNAL EROSION ===")
def calculate_erosion_rate(velocity, sand_concentration, k_material=1e-9, velocity_exponent=2.5):
    """
    Erosion rate estimation (mm/year) based on empirical erosion model:
    E = k_material * C_sand * v^(n)
    """
    return k_material * sand_concentration * (velocity ** velocity_exponent)

# Inputs for verification
v_actual = 25.0             # Fluid/particle velocity in m/s
C_sand_actual = 150.0       # Sand concentration in ppm (mg/L)
k_mat = 2.0e-9              # Material-specific empirical erosion constant
n_exp = 2.5                 # Velocity exponent

# Hand Calculation Verification:
# E = 2e-9 * 150 * (25 ^ 2.5)
# 25 ^ 2.5 = (25 ^ 2) * sqrt(25) = 625 * 5 = 3125
# E = 2e-9 * 150 * 3125 = 3e-7 * 3125 = 0.0009375 mm/year
hand_calc_erosion = 0.0009375
model_calc_erosion = calculate_erosion_rate(v_actual, C_sand_actual, k_material=k_mat, velocity_exponent=n_exp)

print(f"Inputs: Velocity = {v_actual} m/s, Sand Conc = {C_sand_actual} ppm, k = {k_mat}, exponent = {n_exp}")
print(f"Model Calculated Erosion Rate: {model_calc_erosion:.7f} mm/year")
print(f"Hand Calculated Erosion Rate:  {hand_calc_erosion:.7f} mm/year")
assert np.isclose(model_calc_erosion, hand_calc_erosion), "Erosion calculation mismatch!"
print("Validation: PASS\n")


print("=== PHYSICS MODEL 2: PWV SEAL HYDRAULIC LEAKAGE ===")
def calculate_seal_leakage(diameter, clearance, delta_p, viscosity, length):
    """
    Volumetric leak rate (m^3/s) through an annular seal clearance using the slot flow equation:
    Q = (pi * d * c^3 * delta_P) / (12 * mu * L)
    """
    numerator = np.pi * diameter * (clearance ** 3) * delta_p
    denominator = 12 * viscosity * length
    return numerator / denominator

# Inputs for verification
d_seal = 0.050              # Seal nominal diameter: 50 mm (0.050 m)
c_gap = 1.0e-4              # Worn radial clearance gap: 0.1 mm (1e-4 m)
delta_P = 1.5e7             # Differential control pressure: 150 bar (1.5e7 Pa)
mu_fluid = 0.045            # Hydraulic oil dynamic viscosity: 45 cP (0.045 Pa*s)
L_seal = 0.010              # Seal land length: 10 mm (0.010 m)

# Hand Calculation Verification:
# numerator = pi * 0.050 * (1e-4 ^ 3) * 1.5e7 = pi * 0.050 * 1e-12 * 1.5e7 = pi * 0.050 * 1.5e-5 = 7.5e-7 * pi
# denominator = 12 * 0.045 * 0.010 = 0.0054
# Q = (7.5e-7 * pi) / 0.0054 = (2.35619449e-6) / 0.0054 = 0.000436332 m^3/s
hand_calc_leakage = (7.5e-7 * np.pi) / 0.0054
model_calc_leakage = calculate_seal_leakage(d_seal, c_gap, delta_P, mu_fluid, L_seal)

print(f"Inputs: Dia = {d_seal} m, Gap = {c_gap} m, Delta P = {delta_P} Pa, Viscosity = {mu_fluid} Pa*s, Length = {L_seal} m")
print(f"Model Calculated Leakage Rate: {model_calc_leakage:.8f} m^3/s")
print(f"Hand Calculated Leakage Rate:  {hand_calc_leakage:.8f} m^3/s")
assert np.isclose(model_calc_leakage, hand_calc_leakage), "Leakage calculation mismatch!"
print("Validation: PASS")
```

Question 4. Create a risk matrix plot showing failure modes by severity vs. occurrence, with RPN values as annotations.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the aesthetic style
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(11, 8))

# Create a scatter plot representing the failure modes
scatter = ax.scatter(
    x=df_fmeca["O"],
    y=df_fmeca["S"],
    s=df_fmeca["RPN"] * 10,  # Bubble size scaled by RPN
    c=df_fmeca["RPN"],       # Color mapped to RPN value
    cmap="YlOrRd",            # Yellow-Orange-Red gradient
    alpha=0.85,
    edgecolors="black",
    linewidths=1.5
)

# Add annotations for each failure mode with its RPN
for idx, row in df_fmeca.iterrows():
    # Abbreviate the item/subsystem name if too long for cleaner plotting
    label = f"{row['Item/Subsystem']}\n(RPN: {row['RPN']})"
    ax.annotate(
        label,
        (row["O"], row["S"]),
        textcoords="offset points",
        xytext=(12, -5),        # Position offset to avoid overlapping the scatter point
        ha='left',
        va='center',
        fontsize=9,
        fontweight='semibold',
        bbox=dict(boxstyle="round,pad=0.3", fc="white", edgecolor="gray", alpha=0.85, lw=0.5)
    )

# Set labels and range for a standard 1-10 risk rating scale
ax.set_title("Subsea Christmas Tree FMECA Risk Matrix", fontsize=14, fontweight="bold", pad=15)
ax.set_xlabel("Occurrence (O) Rating (1 = Extremely Remote, 10 = High Probability)", fontsize=11, labelpad=10)
ax.set_ylabel("Severity (S) Rating (1 = Minor Impact, 10 = Catastrophic Impact)", fontsize=11, labelpad=10)
ax.set_xlim(0, 6)   # Scaled to focus on the active data range
ax.set_ylim(4, 11)
ax.set_xticks(range(1, 7))
ax.set_yticks(range(4, 12))

# Add a colorbar indicating RPN scales
cbar = fig.colorbar(scatter, ax=ax, shrink=0.75, pad=0.08)
cbar.set_label("Risk Priority Number (RPN)", rotation=270, labelpad=15, fontsize=10, fontweight="bold")

plt.tight_layout()
plt.show()
```
